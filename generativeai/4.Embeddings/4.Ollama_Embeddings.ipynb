### Ollama Local Embeddings (OllamaEmbeddings)

Ollama allows running open-source LLMs and embedding models locally via a self-hosted REST daemon.

Models such as nomic-embed-text and mxbai-embed-large provide excellent retrieval benchmarks for local RAG pipelines.

### Setup & Prerequisites

1. Install Ollama daemon from [ollama.com](https://ollama.com/).
2. Pull target embedding model: ollama pull nomic-embed-text.
3. Ensure server is running at default base_url: http://localhost:11434.

```python
from langchain_community.embeddings import OllamaEmbeddings

try:
    ollama_embeddings = OllamaEmbeddings(
        model="nomic-embed-text",
        base_url="http://localhost:11434"
    )

    doc_text = "Ollama enables local model execution with zero cloud API dependencies."
    vector = ollama_embeddings.embed_query(doc_text)
    print(f"Ollama Embedding Vector Length: {len(vector)} dimensions")
    print(f"Sample vector values: {vector[:5]}")
except Exception as e:
    print("Ollama daemon not reachable at http://localhost:11434 or model not pulled.")
    print(f"Details: {e}")
    print("\nSetup steps:")
    print("1. Start Ollama: 'ollama serve'")
    print("2. Pull model: 'ollama pull nomic-embed-text'")
    print("\nAPI Code Structure:")
    print('embeddings = OllamaEmbeddings(model="nomic-embed-text")')
```
