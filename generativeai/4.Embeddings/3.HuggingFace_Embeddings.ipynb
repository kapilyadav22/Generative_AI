### HuggingFace Local Embeddings (HuggingFaceEmbeddings)

HuggingFaceEmbeddings runs embedding models entirely offline on your local CPU or GPU using PyTorch and sentence-transformers.

This approach guarantees 100% data privacy, eliminates API costs, and avoids external network latency during RAG pipeline execution.

### Recommended Open-Source Models

* all-MiniLM-L6-v2: Extremely fast, lightweight model (384 dimensions).
* BAAI/bge-small-en-v1.5: High-accuracy retrieval model (384 dimensions).
* BAAI/bge-large-en-v1.5: State-of-the-art embedding model (1024 dimensions).

```python
import numpy as np
from langchain_huggingface import HuggingFaceEmbeddings

# Initialize HuggingFace embeddings running locally on CPU
hf_embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)

docs = [
    "HuggingFace provides open-source transformer models for NLP tasks.",
    "Local embeddings allow offline RAG without sending data to cloud APIs."
]
query = "Can I run embeddings without an API key?"

doc_vecs = hf_embeddings.embed_documents(docs)
query_vec = hf_embeddings.embed_query(query)

print(f"HuggingFace Vector Dimension: {len(doc_vecs[0])}")
print(f"Query Vector Length: {len(query_vec)}")

sim_1 = np.dot(query_vec, doc_vecs[0])
sim_2 = np.dot(query_vec, doc_vecs[1])

print(f"Similarity to Doc 1: {sim_1:.4f}")
print(f"Similarity to Doc 2: {sim_2:.4f}")
```
