### Azure OpenAI Embeddings (AzureOpenAIEmbeddings)

Azure OpenAI Service provides enterprise-grade OpenAI models hosted on Microsoft Azure infrastructure.

It offers enterprise security, regional availability, private endpoint integration (VNet), and guaranteed compliance (SOC2/HIPAA) for corporate cloud applications.

### Key Configuration Parameters

* azure_deployment: Deployment name created in Azure AI Studio.
* openai_api_version: API version (e.g. 2024-02-01).
* azure_endpoint: Base URL endpoint (https://<your-resource-name>.openai.azure.com/).
* api_key / AZURE_OPENAI_API_KEY: Azure API key.

```python
import os
from dotenv import load_dotenv
from langchain_openai import AzureOpenAIEmbeddings

load_dotenv()

azure_key = os.getenv("AZURE_OPENAI_API_KEY")
azure_endpoint = os.getenv("AZURE_OPENAI_ENDPOINT")
azure_deployment = os.getenv("AZURE_OPENAI_DEPLOYMENT_NAME", "text-embedding-ada-002")
api_version = os.getenv("AZURE_OPENAI_API_VERSION", "2024-02-01")

if azure_key and azure_endpoint:
    embeddings = AzureOpenAIEmbeddings(
        azure_deployment=azure_deployment,
        openai_api_version=api_version,
        azure_endpoint=azure_endpoint,
        api_key=azure_key
    )

    vector = embeddings.embed_query("Enterprise cloud security and compliance.")
    print(f"Azure OpenAI Embedding Vector Length: {len(vector)} dimensions")
    print(f"Sample vector values: {vector[:5]}")
else:
    print("Azure OpenAI environment variables not found.")
    print("To configure Azure OpenAI, set the following in your .env:")
    print("AZURE_OPENAI_API_KEY=your_key")
    print("AZURE_OPENAI_ENDPOINT=https://your-resource.openai.azure.com/")
    print("AZURE_OPENAI_DEPLOYMENT_NAME=your_deployment_name")
    print("AZURE_OPENAI_API_VERSION=2024-02-01")
    print("\nAPI Code Structure:")
    print('embeddings = AzureOpenAIEmbeddings(')
    print('    azure_deployment="text-embedding-3-small",')
    print('    openai_api_version="2024-02-01",')
    print('    azure_endpoint="https://your-resource.openai.azure.com/"')
    print(')')
```
