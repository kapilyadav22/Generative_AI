### Google Gemini Embeddings (GoogleGenerativeAIEmbeddings)

Google Gemini Embeddings (models/gemini-embedding-001) generate high-dimensional (3072-dim) vector representations.

It supports explicit task_type optimization parameters to tailor vector spaces specifically for document indexing versus query retrieval.

### Key Parameters & Task Types

* model: Target model name (e.g. gemini-embedding-001).
* task_type: Optimization task type (RETRIEVAL_DOCUMENT, RETRIEVAL_QUERY, SEMANTIC_SIMILARITY, CLASSIFICATION).
* google_api_key / GOOGLE_API_KEY: Authentication key loaded from .env.

```python
import os
import numpy as np
from dotenv import load_dotenv
from langchain_google_genai import GoogleGenerativeAIEmbeddings

load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")

if api_key:
    embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")

    documents = [
        "Quantum computing uses qubits and superposition to perform parallel computations.",
        "French cuisine is famous for rich sauces, fresh baguettes, and fine wine."
    ]
    query = "How does quantum computing work?"

    doc_embeddings = embeddings.embed_documents(documents)
    query_embedding = embeddings.embed_query(query)

    print(f"Gemini Embedding Vector Dimension: {len(doc_embeddings[0])}")
    print(f"Sample Document Vector (first 5 elements): {doc_embeddings[0][:5]}")

    # Calculate Cosine Similarity
    q_vec = np.array(query_embedding)
    d1_vec = np.array(doc_embeddings[0])
    d2_vec = np.array(doc_embeddings[1])

    sim_1 = np.dot(q_vec, d1_vec) / (np.linalg.norm(q_vec) * np.linalg.norm(d1_vec))
    sim_2 = np.dot(q_vec, d2_vec) / (np.linalg.norm(q_vec) * np.linalg.norm(d2_vec))

    print(f"Similarity to Quantum Doc: {sim_1:.4f}")
    print(f"Similarity to Cuisine Doc: {sim_2:.4f}")
else:
    print("GOOGLE_API_KEY not set in environment.")
```
