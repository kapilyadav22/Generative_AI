### OpenAI Embeddings (OpenAIEmbeddings)

OpenAI provides high-performance embedding models including text-embedding-3-small and text-embedding-3-large.

The text-embedding-3 family supports **Matryoshka representation learning**, allowing developers to shorten vector dimensions (e.g. from 1536 down to 512) to save storage without losing significant retrieval accuracy.

### Key Parameters

* model: Model name (default text-embedding-3-small).
* dimensions: Optional integer to reduce embedding vector length.
* api_key / OPENAI_API_KEY: Authentication key loaded from environment.

```python
import os
import numpy as np
from dotenv import load_dotenv
from langchain_openai import OpenAIEmbeddings

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

if api_key:
    # Initialize OpenAI Embeddings with reduced dimensions (512)
    embeddings = OpenAIEmbeddings(
        model="text-embedding-3-small",
        dimensions=512
    )

    docs = [
        "LangChain simplifies LLM application development.",
        "Vector databases store dense embeddings for fast similarity search."
    ]
    query = "What is a vector database?"

    doc_vectors = embeddings.embed_documents(docs)
    query_vector = embeddings.embed_query(query)

    print(f"Document 1 Vector Length: {len(doc_vectors[0])} dimensions")
    print(f"Query Vector Length: {len(query_vector)} dimensions")
    print(f"Sample vector values: {doc_vectors[0][:5]}")
else:
    print("OPENAI_API_KEY not found in environment.")
    print("To execute OpenAI embeddings, set OPENAI_API_KEY in your .env file.")
    print("API usage example:")
    print('embeddings = OpenAIEmbeddings(model="text-embedding-3-small", dimensions=512)')
```
