# Vector Embeddings: Core Concepts & Architecture

**Vector Embeddings** transform unstructured text (words, sentences, documents) into dense floating-point numerical vectors in a high-dimensional continuous vector space.

In modern Large Language Model (LLM) and Retrieval-Augmented Generation (RAG) architectures, embeddings enable semantic search by ensuring texts with similar meanings are located close to each other in vector space, even if they use completely different vocabulary.

### Distance & Similarity Metrics

To evaluate how semantically related two text embeddings $\mathbf{u}$ and $\mathbf{v}$ are, vector databases and frameworks compute distance metrics:

1. **Cosine Similarity**: Measures the cosine of the angle between two vectors. Focuses purely on direction rather than vector magnitude.
   $$\text{Cosine Similarity}(\mathbf{u}, \mathbf{v}) = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\| \|\mathbf{v}\|}$$

2. **Euclidean Distance ($L_2$)**: Measures straight-line geometric distance between two vector endpoints in space.
   $$d(\mathbf{u}, \mathbf{v}) = \sqrt{\sum_{i=1}^{n} (u_i - v_i)^2}$$

3. **Dot Product (Inner Product)**: Measures direction and magnitude. When vectors are normalized to unit length ($\|\mathbf{u}\| = 1$), dot product equals cosine similarity.
   $$\text{Dot Product}(\mathbf{u}, \mathbf{v}) = \mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^{n} u_i v_i$$

### Document vs. Query Embeddings

LangChain embedding classes provide two primary interface methods:

* embed_documents(texts: List[str]) -> List[List[float]]: Embeds multiple background documents for storage and indexing in a vector database.
* embed_query(text: str) -> List[float]: Embeds a single search query prompt for retrieving relevant document vectors.

```python
import numpy as np

# Demonstrating distance metrics calculation using numpy
vec_a = np.array([0.2, 0.8, 0.5])
vec_b = np.array([0.25, 0.75, 0.52])
vec_c = np.array([-0.9, 0.1, -0.2])

def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

def euclidean_distance(u, v):
    return np.linalg.norm(u - v)

print("--- Vector Similarity Computations ---")
print(f"Cosine Sim (A vs B - similar): {cosine_similarity(vec_a, vec_b):.4f}")
print(f"Cosine Sim (A vs C - different): {cosine_similarity(vec_a, vec_c):.4f}")
print(f"Euclidean Dist (A vs B): {euclidean_distance(vec_a, vec_b):.4f}")
print(f"Euclidean Dist (A vs C): {euclidean_distance(vec_a, vec_c):.4f}")
```
