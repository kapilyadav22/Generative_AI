# Basic Message History Management (ChatMessageHistory)

This notebook demonstrates direct message history operations using ChatMessageHistory.

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

# Initialize in-memory chat message history
history = ChatMessageHistory()

# Add User and AI messages
history.add_user_message("Hi, my name is Kapil.")
history.add_ai_message("Hello Kapil! How can I help you today?")
history.add_user_message("I am learning LangChain Memory concepts.")
history.add_ai_message("Memory allows LLMs to retain conversation history across turns.")

print(f"Total messages in history: {len(history.messages)}\n")

print("=== Inspecting Stored Messages ===")
for msg in history.messages:
    print(f"[{type(msg).__name__}]: {msg.content}")

history.clear()
print(f"\nMessages after clear(): {len(history.messages)}")
```
