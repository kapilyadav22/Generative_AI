# Memory + Agents Integration

Integrates modern LCEL conversational memory into a Tool-Calling Agent using ChatGoogleGenerativeAI (gemini-3.6-flash), MessagesPlaceholder for chat_history, and AgentExecutor.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_core.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_community.chat_message_histories import ChatMessageHistory

load_dotenv(find_dotenv())

@tool
def multiply(a: float, b: float) -> float:
    """Multiply two numbers together."""
    return a * b

tools = [multiply]
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an agent with calculation tools and conversation memory."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

agent = create_tool_calling_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)

# Modern Conversational Agent Execution Helper
history = ChatMessageHistory()

def ask_conversational_agent(user_query: str):
    res = agent_executor.invoke({"input": user_query, "chat_history": history.messages})
    history.add_user_message(user_query)
    history.add_ai_message(res["output"])
    return res["output"]

print("Memory + Agent pipeline initialized successfully with gemini-3.6-flash!")
```
