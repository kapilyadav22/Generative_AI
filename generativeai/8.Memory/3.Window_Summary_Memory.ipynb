# Context Window Management with trim_messages

Demonstrates using trim_messages from langchain_core.messages to enforce sliding window token/message limits.

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage, trim_messages

messages = [
    SystemMessage(content="You are an AI assistant."),
    HumanMessage(content="Hi, my favorite color is Blue."),
    AIMessage(content="That's nice! Blue is a soothing color."),
    HumanMessage(content="My favorite programming language is Python."),
    AIMessage(content="Python is versatile and clean."),
    HumanMessage(content="My favorite framework is LangChain."),
    AIMessage(content="LangChain simplifies LLM application development."),
    HumanMessage(content="What was my favorite color?")
]

print(f"Original Message Count: {len(messages)}")

trimmed = trim_messages(
    messages,
    max_tokens=4,
    token_counter=len,
    strategy="last",
    include_system=True,
    allow_partial=False
)

print(f"Trimmed Message Count: {len(trimmed)}\n")
print("=== Retained Messages in Context Window ===")
for msg in trimmed:
    print(f"[{type(msg).__name__}]: {msg.content}")
```
