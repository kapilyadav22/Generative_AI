# Memory in LangChain: Architecture & Concepts

By default, Large Language Models are stateless. Every request to an LLM API is executed independently without inherent memory of previous turns.

**Memory** in LangChain is the abstraction responsible for persisting, formatting, and inserting conversation history into prompt context across multi-turn interactions.

### Short-Term Memory vs. Long-Term Memory

1. **Short-Term Memory (In-Session / Working Memory)**:
   - Scope: Active single-session chat duration.
   - Storage: In-memory list (InMemoryChatMessageHistory, ChatMessageHistory).
   - Lifecycle: Flushed when application restarts or session terminates.
   - Goal: Retains immediate conversational turns for coherence during a chat.

2. **Long-Term Memory (Cross-Session / Persistent Memory)**:
   - Scope: Cross-session, multi-day, or user profile persistence.
   - Storage: External databases (Redis, PostgreSQL, FileChatMessageHistory, VectorStores).
   - Lifecycle: Persists indefinitely across user logins and restarts.
   - Goal: Enables user preference tracking, historical memory retrieval, and personalized experiences.

### Modern LCEL Memory Pattern (Replacing Deprecated RunnableWithMessageHistory)

Legacy RunnableWithMessageHistory is deprecated in favor of explicit LCEL history passing.

In modern LCEL:
1. MessagesPlaceholder("chat_history") is declared in ChatPromptTemplate.
2. ChatMessageHistory (or FileChatMessageHistory) manages message state.
3. History is passed directly at call time: chain.invoke({"input": query, "chat_history": history.messages}).
4. User questions and AI answers are stored explicitly in history, offering 100% control over state.

### Comprehensive Memory Types Comparison Matrix

| Memory Pattern | Storage Mechanism | Primary Behavior | Ideal Use Case |
|---|---|---|---|
| **ChatMessageHistory** | In-Memory List | Stores raw sequence of HumanMessage and AIMessage objects | Short-session chatbots, prototyping |
| **Window Memory (trim_messages)** | Sliding Window / Token Limit | Retains last k messages or last N tokens, dropping oldest turns | Long conversations with strict LLM context limits |
| **FileChatMessageHistory** | Disk JSON File | Persists chat turns to disk files | Local apps needing session recovery across restarts |
| **Redis / Postgres Memory** | Database | Persists session histories into external production databases | Enterprise scalable multi-tenant applications |
| **VectorStore Memory** | Vector Database | Stores history as embeddings for semantic similarity retrieval | Retrieving relevant past memories from months ago |
