# Short-Term Memory vs. Long-Term Memory

Demonstrates the practical difference between:
* Short-Term Memory: Ephemeral InMemoryChatMessageHistory (flushed when process restarts).
* Long-Term Memory: FileChatMessageHistory (persisted to disk as JSON, reloaded across restarts).

```python
import os
import json
from langchain_community.chat_message_histories import ChatMessageHistory, FileChatMessageHistory

# 1. Short-Term Memory (Ephemeral in-memory list)
short_memory = ChatMessageHistory()
short_memory.add_user_message("Transient message for current session.")
print(f"Short-Term Memory Count: {len(short_memory.messages)}")

# 2. Long-Term Memory (Persistent JSON file storage)
store_dir = "./memory_store"
os.makedirs(store_dir, exist_ok=True)
file_path = os.path.join(store_dir, "user_session_101.json")

long_memory = FileChatMessageHistory(file_path)
long_memory.add_user_message("Persistent user preference: Loves Python and Gemini 3.6 Flash.")
long_memory.add_ai_message("Saved preference to persistent disk storage.")

print(f"Long-Term Memory Count: {len(long_memory.messages)}")
print(f"Disk File Exists: {os.path.exists(file_path)}")

# Reload from disk
reloaded_memory = FileChatMessageHistory(file_path)
print("\n=== Reloaded Long-Term Memory from Disk ===")
for msg in reloaded_memory.messages:
    print(f"[{type(msg).__name__}]: {msg.content}")

# Clean up file store after demo
if os.path.exists(file_path):
    os.remove(file_path)
if os.path.exists(store_dir):
    os.rmdir(store_dir)
```
