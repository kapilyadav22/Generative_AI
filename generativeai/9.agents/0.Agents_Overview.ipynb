# AI Agents in LangChain: Architecture & Principles

An **Agent** in LangChain is an autonomous system that uses a Language Model as a reasoning engine to dynamically decide a sequence of actions, execute tools, observe results, and iterate until completing a task.

### Chains vs. Agents

1. **Chains**: Hardcoded execution pipelines defined entirely by the developer at build time. The sequence of steps (e.g. Prompt -> LLM -> Parser) never changes.
2. **Agents**: Dynamic execution systems where the LLM evaluates the user input, decides which tools to call, inspects tool observations, and chooses subsequent actions at runtime.

### The Agent Perception-Action Execution Loop

1. **Perception / Input**: The user sends a prompt alongside conversation history.
2. **Reasoning (Thought)**: The LLM analyzes the user goal and available tools, determining whether a tool invocation is required.
3. **Action Selection**: The LLM outputs a structured tool call request specifying tool name and argument values.
4. **Tool Execution**: The AgentExecutor runtime executes the corresponding Python function or API.
5. **Observation**: Tool execution output (e.g. calculation result or database record) is appended to agent scratchpad context.
6. **Iteration / Completion**: The loop repeats until the LLM produces a final answer.

### Core Agent Components

* **Reasoning Engine (LLM)**: Model capable of function/tool calling (e.g. Google Gemini 2.0 Flash).
* **Tools (@tool)**: Utility functions decorated with clean docstrings describing purpose and input schemas.
* **Agent Prompt**: System prompt template containing placeholders for input, chat history, and agent scratchpad.
* **AgentExecutor**: The runtime loop controller managing iteration limits, error handling, and tool execution.

### Agent Types Comparison

| Agent Type | Component / Function | Mechanism | Best Use Case |
|---|---|---|---|
| **Tool-Calling Agent** | create_tool_calling_agent | Native model function calling API | Modern LLMs (Gemini, OpenAI, Anthropic) |
| **ReAct Agent** | create_react_agent | Prompt-based Thought/Action/Observation text loop | Legacy models without native tool calling |
| **Structured Chat Agent** | create_structured_chat_agent | Multi-argument JSON schema validation | Complex APIs requiring structured JSON inputs |

### Production Best Practices & Guardrails

* Set max_iterations (e.g. max_iterations=5) to prevent infinite loops.
* Write explicit, detailed tool docstrings so the LLM correctly understands when to select each tool.
* Enable handle_parsing_errors=True in AgentExecutor to recover gracefully from schema output issues.
