# Basic AI Agent Creation in LangChain

This notebook demonstrates the absolute simplest way to build an AI Agent in LangChain using a basic custom tool, ChatGoogleGenerativeAI, and AgentExecutor.

### Step 1: Define a Simple Tool

A tool is a standard Python function decorated with `@tool`. The docstring tells the LLM what the tool does.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_core.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

load_dotenv(find_dotenv())

# Define a simple tool
@tool
def multiply(a: int, b: int) -> int:
    """Multiply two integers and return the product."""
    return a * b

tools = [multiply]
print(f"Created tool: {multiply.name}")
```

### Step 2: Initialize LLM & Create Simple Agent

We construct a basic prompt with `agent_scratchpad` and build the agent using `create_tool_calling_agent`.

```python
# Initialize Gemini LLM
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash", temperature=0)

# Define minimal agent prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# Create agent and executor
agent = create_tool_calling_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)

print("Simple Agent created successfully!")
```

### Step 3: Run the Agent

Invoke the agent with a prompt that triggers the tool.

```python
result = agent_executor.invoke({"input": "What is 12 multiplied by 8?"})
print(f"\nAgent Answer: {result['output']}")
```
