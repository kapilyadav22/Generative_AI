# Step 1: Your First 5-Line Middleware (Beginner Friendly)

This notebook builds the absolute simplest middleware that logs every time Gemini 3.6 Flash starts and finishes generating a response.

### Creating a Simple Middleware Handler

We inherit from BaseCallbackHandler and define two simple methods: on_llm_start and on_llm_end.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_core.callbacks import BaseCallbackHandler
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv(find_dotenv())

# 1. Create a simple Middleware Handler
class MyFirstMiddleware(BaseCallbackHandler):
    def on_llm_start(self, serialized, prompts, **kwargs):
        print("\n🟢 [MIDDLEWARE] Starting LLM Call: Sending prompt to Gemini...")
    
    def on_llm_end(self, response, **kwargs):
        print("🔴 [MIDDLEWARE] LLM Call Finished: Received response from Gemini!\n")

# 2. Attach the middleware to Gemini 3.6 Flash
my_middleware = MyFirstMiddleware()
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash", callbacks=[my_middleware])

# 3. Test the middleware with a simple invocation
response = llm.invoke("What is the capital of France?")
print(f"AI Response: {response.content}")
```
