### CharacterTextSplitter

CharacterTextSplitter splits text based on a single user-defined character separator (default is "\n\n").

It measures length based on character count (or custom length function) and enforces chunk_size and chunk_overlap limits.

### Key Parameters

* separator: Character sequence used to split text (default is "\n\n").
* chunk_size: Maximum character count allowed per chunk.
* chunk_overlap: Number of characters shared between adjacent chunks to maintain context.
* is_separator_regex: Set to True if separator is a regex pattern.
* length_function: Function to measure text length (default is len).

```python
from langchain_text_splitters import CharacterTextSplitter
from langchain_core.documents import Document

# Sample raw text payload
sample_text = """LangChain is a framework for developing applications powered by LLMs.

Document Loaders retrieve data from various sources as Document objects.

Text Splitters break down large documents into smaller chunks suitable for embedding.

Vector Stores index and retrieve document chunks using similarity search.
"""

# Initialize CharacterTextSplitter
splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=120,
    chunk_overlap=20,
    is_separator_regex=False
)

# 1. Split raw text string
text_chunks = splitter.split_text(sample_text)
print(f"Total text chunks: {len(text_chunks)}")
for i, chunk in enumerate(text_chunks, 1):
    print(f"--- Chunk {i} ({len(chunk)} chars) ---\n{chunk}\n")

# 2. Split LangChain Document objects
doc = Document(page_content=sample_text, metadata={"source": "overview.txt"})
doc_chunks = splitter.split_documents([doc])
print(f"Total Document chunks: {len(doc_chunks)}")
print(f"Chunk 1 Metadata: {doc_chunks[0].metadata}")
```
