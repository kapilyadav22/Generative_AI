### SemanticChunker (Embedding-Based Splitting)

SemanticChunker splits text based on semantic meaning rather than fixed character or token counts.

It splits text into sentences, computes embedding vectors for adjacent sentences, and places a chunk split boundary whenever semantic distance between consecutive sentences exceeds a defined threshold.

### Breakpoint Threshold Types

* percentile (default): Splits when semantic distance exceeds a percentile threshold (e.g., 95th percentile of differences).
* standard_deviation: Splits when difference exceeds N standard deviations from mean similarity.
* interquartile: Splits based on interquartile range (IQR) distance boundaries.
* gradient: Evaluates semantic distance gradient changes across consecutive sentence windows.

```python
import os
from dotenv import load_dotenv
from langchain_experimental.text_splitter import SemanticChunker
from langchain_google_genai import GoogleGenerativeAIEmbeddings

load_dotenv()

# Initialize embedding model for semantic distance evaluation
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")

# Sample text with distinct semantic topic transitions
multi_topic_text = """Quantum computing uses qubits to perform parallel computations leveraging superposition and entanglement.
Unlike classical bits which are binary 0 or 1, qubits represent complex quantum states.

French cuisine is renowned worldwide for its rich sauces, pastries, and culinary techniques.
Famous dishes include coq au vin, boeuf bourguignon, and croissants.

Modern astronomy relies on space telescopes like James Webb to observe distant galaxies and cosmic microwave background radiation.
Astronomers analyze light spectra to identify elemental compositions of exoplanets.
"""

# Initialize SemanticChunker with percentile breakpoint threshold
semantic_splitter = SemanticChunker(
    embeddings=embeddings,
    breakpoint_threshold_type="percentile",
    breakpoint_threshold_amount=90.0
)

# Split text into semantic chunks
semantic_chunks = semantic_splitter.split_text(multi_topic_text)

print(f"Generated {len(semantic_chunks)} semantic chunks based on topic shifts:")
for i, chunk in enumerate(semantic_chunks, 1):
    print(f"--- Semantic Chunk {i} ---\n{chunk.strip()}\n")
```
