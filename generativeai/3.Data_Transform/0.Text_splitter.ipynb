### Text Splitters in LangChain

Text Splitters break large documents into smaller, semantically coherent chunks suitable for embedding and retrieval in RAG pipelines.

When ingesting long documents (PDFs, raw text, scraped web pages, code files), passing entire documents directly to an LLM or embedding model is inefficient or impossible due to context window limits.

---

### Why Text Splitting is Critical in RAG

* Context Window Constraints: LLMs have token input limits (e.g., 4k, 8k, 32k tokens). Text splitters prevent prompt overflow errors.
* Embedding Accuracy: Dense vector models generate sharper, more precise embeddings on small, focused text chunks rather than massive multi-page documents.
* Retrieval Granularity: Smaller chunks allow vector databases to retrieve only the exact relevant passages needed to answer a user prompt.

---

### Key Parameters Across Text Splitters

* chunk_size: Maximum size (characters or tokens) of each generated text chunk.
* chunk_overlap: Number of characters/tokens shared between adjacent chunks. Overlap prevents information loss at chunk boundaries.
* length_function: Function used to calculate chunk length (default is len for characters; tiktoken or custom functions for tokens).
* separators: Ordered list of character sequences used to locate natural split boundaries (e.g., ["\n\n", "\n", " ", ""]).

---

### Standard Splitter Interface

* split_text(text: str) -> List[str]: Accepts a raw string payload and returns a list of text chunks.
* split_documents(documents: List[Document]) -> List[Document]: Accepts a list of LangChain Document objects, splits page_content, and preserves metadata across all generated sub-chunks.
