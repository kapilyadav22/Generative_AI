### CodeTextSplitter (Language-Aware Splitting)

CodeTextSplitter uses programming language AST boundaries (class declarations, function definitions, control blocks) to split source code files without breaking code logic.

It is constructed using RecursiveCharacterTextSplitter.from_language() specifying a target language enum.

### Supported Languages

Supported languages include Python, JavaScript, TypeScript, HTML, Markdown, C++, Java, Go, Rust, PHP, Ruby, Swift, SQL, and more.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language

# Sample Python script content
python_code = """class DatabaseConnection:
    def __init__(self, uri: str):
        self.uri = uri
        self.connected = False

    def connect(self):
        print(f"Connecting to {self.uri}...")
        self.connected = True
        return self.connected

def calculate_total(prices: list[float]) -> float:
    total = sum(prices)
    return total

def main():
    db = DatabaseConnection("postgres://localhost:5432/db")
    db.connect()
    print(f"Total: {calculate_total([10.5, 20.0, 5.25])}")
"""

# Create Python-aware code splitter
python_splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=150,
    chunk_overlap=20
)

code_chunks = python_splitter.split_text(python_code)
print(f"Generated {len(code_chunks)} code chunks:")
for i, chunk in enumerate(code_chunks, 1):
    print(f"--- Code Chunk {i} ---\n{chunk}\n")
```
