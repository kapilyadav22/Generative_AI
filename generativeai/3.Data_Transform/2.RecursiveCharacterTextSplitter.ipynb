### RecursiveCharacterTextSplitter

RecursiveCharacterTextSplitter is the recommended default splitter for generic text in LangChain.

It attempts to split text using a hierarchical list of separators (default: ["\n\n", "\n", " ", ""]) in order, keeping paragraphs, sentences, and words together as long as possible.

### Why Use RecursiveCharacterTextSplitter?

* Preserves Semantic Boundaries: Tries paragraph splits first, then sentence splits, then word splits.
* Prevents Fragmented Chunks: Ensures words and sentences are not arbitrarily sliced mid-character.
* Highly Configurable: Allows custom separator hierarchy and custom length functions.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

# Sample long article
article = """Retrieval-Augmented Generation (RAG) improves LLM factual accuracy.

Instead of relying only on static training weights, RAG retrieves relevant documents from an external vector store.
The retrieved passages are provided alongside the prompt context to the chat model.

Text splitting is a critical pre-processing step in RAG pipelines. If chunks are too large, irrelevant context lowers retrieval quality. If chunks are too small, key semantic context is lost.
"""

# Initialize RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=30,
    separators=["\n\n", "\n", " ", ""]
)

# 1. Split raw text
chunks = splitter.split_text(article)
print(f"Total chunks generated: {len(chunks)}")
for i, c in enumerate(chunks, 1):
    print(f"--- Chunk {i} ({len(c)} chars) ---\n{c}\n")

# 2. Split Document objects with metadata
doc = Document(page_content=article, metadata={"topic": "RAG Architecture", "author": "AI Team"})
doc_splits = splitter.split_documents([doc])
print(f"Document splits generated: {len(doc_splits)}")
print(f"Split 1 Metadata: {doc_splits[0].metadata}")
```
