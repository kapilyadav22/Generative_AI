### RecursiveJsonSplitter (JSON-Aware Splitting)

RecursiveJsonSplitter recursively traverses nested JSON objects and arrays to split large JSON structures into smaller JSON chunks while preserving valid JSON hierarchy.

It ensures key-value context is maintained without arbitrarily slicing across JSON key-value boundaries.

### Key Methods & Features

* split_json(json_data): Returns chunks as a list of Python dictionaries / JSON objects.
* split_text(json_data): Returns chunks formatted as valid JSON strings.
* create_documents(texts): Converts JSON chunks into LangChain Document objects.
* max_chunk_size: Maximum character length allowed for each split JSON chunk (default is 2000).
* min_chunk_size: Minimum character size threshold before combining small nested JSON elements.

```python
import json
from langchain_text_splitters import RecursiveJsonSplitter

# Sample nested JSON dataset
nested_json_data = {
    "company": "TechCorp Global",
    "departments": [
        {
            "name": "Engineering",
            "head": "Sarah Jenkins",
            "teams": [
                {"name": "AI Core", "lead": "Alex Rivera", "projects": ["RAG Engine", "Agentic Framework"]},
                {"name": "Infrastructure", "lead": "David Chen", "projects": ["K8s Migration", "Vector DB Mesh"]}
            ]
        },
        {
            "name": "Product & Design",
            "head": "Elena Rostova",
            "teams": [
                {"name": "UX Research", "lead": "Marcus Vance", "projects": ["User Journey Map", "Accessibility Audit"]}
            ]
        }
    ],
    "locations": ["San Francisco", "New York", "London", "Tokyo"]
}

# Initialize RecursiveJsonSplitter with max_chunk_size limit
json_splitter = RecursiveJsonSplitter(max_chunk_size=300)

# 1. Split JSON into Python dictionaries
json_chunks = json_splitter.split_json(json_data=nested_json_data)
print(f"Generated {len(json_chunks)} JSON dictionary chunks:")
for i, chunk in enumerate(json_chunks, 1):
    print(f"--- JSON Chunk {i} ---\n{json.dumps(chunk, indent=2)}\n")

# 2. Split JSON directly into LangChain Document objects
documents = json_splitter.create_documents(texts=[nested_json_data])
print(f"Generated {len(documents)} LangChain Document objects:")
for i, doc in enumerate(documents, 1):
    print(f"--- Document {i} ---\n{doc.page_content}\n")
```
