### HTMLHeaderTextSplitter

HTMLHeaderTextSplitter parses HTML documents and splits text at specified HTML header elements (<h1>, <h2>, <h3>, <h4>).

It attaches the HTML header hierarchy to the metadata of each generated document chunk.

### Use Cases

* Scraped Web Content: Processing HTML web pages scraped using WebBaseLoader.
* HTML Documentation: Ingesting HTML-rendered user manuals and technical docs.
* Metadata Attribution: Retaining web section hierarchy in RAG vector stores.

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

# Sample HTML page content
html_string = """<!DOCTYPE html>
<html>
<body>
    <h1>Generative AI Platform</h1>
    <p>Generative AI is revolutionizing business workflows worldwide.</p>
    
    <h2>Core Features</h2>
    <p>Our platform provides LLM orchestration, RAG pipelines, and agent memory.</p>
    
    <h3>Vector Database Integration</h3>
    <p>Store and search high-dimensional embeddings efficiently.</p>
</body>
</html>
"""

# Define HTML headers to split on
headers_to_split_on = [
    ("h1", "Header 1"),
    ("h2", "Header 2"),
    ("h3", "Header 3"),
]

html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
html_splits = html_splitter.split_text(html_string)

print(f"Generated {len(html_splits)} HTML splits:")
for split in html_splits:
    print(f"Metadata: {split.metadata}")
    print(f"Content: {split.page_content.strip()}\n")
```
