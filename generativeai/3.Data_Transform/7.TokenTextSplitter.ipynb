### TokenTextSplitter

TokenTextSplitter splits text based on raw LLM token counts rather than character counts.

Because LLM context limits and API billing pricing are measured strictly in tokens, token-based splitting guarantees that chunks never exceed target token limits.

### Why Token-Based Splitting Matters

* Exact Context Window Matching: Guarantees retrieved chunks fit exactly inside prompt budget allocations.
* Tokenizer Precision: Uses BPE tokenizers (such as cl100k_base or p50k_base via tiktoken).
* Avoids Context Overflow: Eliminates prompt truncation errors caused by variable character-to-token ratios.

```python
from langchain_text_splitters import TokenTextSplitter
from langchain_core.documents import Document

# Sample long text payload
sample_text = """LangChain provides modular components for building context-aware reasoning applications.

TokenTextSplitter uses the tiktoken library under the hood to measure the exact number of BPE tokens in a passage.

By specifying encoding_name='cl100k_base', we ensure that chunk size measurements directly match OpenAI GPT-4 and text-embedding-3 model tokenizers.
"""

# Initialize TokenTextSplitter (e.g., max 25 tokens per chunk)
token_splitter = TokenTextSplitter(
    encoding_name="cl100k_base",
    chunk_size=25,
    chunk_overlap=5
)

# 1. Split raw text by token count
token_chunks = token_splitter.split_text(sample_text)
print(f"Generated {len(token_chunks)} token-based chunks:")
for i, chunk in enumerate(token_chunks, 1):
    print(f"--- Token Chunk {i} ---\n{chunk}\n")

# 2. Split Document objects
doc = Document(page_content=sample_text, metadata={"model": "gpt-4o"})
doc_chunks = token_splitter.split_documents([doc])
print(f"Total Document chunks: {len(doc_chunks)}")
print(f"Chunk 1 Metadata: {doc_chunks[0].metadata}")
```
