### MarkdownHeaderTextSplitter

MarkdownHeaderTextSplitter splits Markdown text by specified header levels (#, ##, ###) and automatically populates header titles into document metadata.

This ensures that retrieved chunks carry their parent section headers as metadata for downstream LLM generation.

### Key Advantages

* Header Context Preservation: Every chunk knows which section and sub-section it belongs to.
* Preserves Markdown Structure: Keeps header sections together logically.
* Combinable: Can be paired with RecursiveCharacterTextSplitter for secondary character-based chunking.

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter, RecursiveCharacterTextSplitter

# Sample Markdown document
markdown_document = """# Building LLM Applications
Large Language Models are transforming software development.

## Data Ingestion
Ingestion converts raw documents into standardized Document objects.

### Document Loaders
Loaders fetch data from local files, PDFs, APIs, or databases.

## Data Transformation
Transforming documents involves text splitting and chunking.

### Text Splitters
Splitters break large texts into optimal chunk sizes for embedding.
"""

# Define headers to split on
headers_to_split_on = [
    ("#", "Header 1"),
    ("##", "Header 2"),
    ("###", "Header 3"),
]

# Initialize splitter
md_splitter = MarkdownHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
md_splits = md_splitter.split_text(markdown_document)

print(f"Generated {len(md_splits)} Markdown splits:")
for split in md_splits:
    print(f"Metadata: {split.metadata}")
    print(f"Content: {split.page_content[:80]}\n")

# Optional: Combine with RecursiveCharacterTextSplitter for long section management
char_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=10)
final_splits = char_splitter.split_documents(md_splits)
print(f"Final sub-splits count: {len(final_splits)}")
print(f"Sub-split 1 Metadata: {final_splits[0].metadata}")
```
