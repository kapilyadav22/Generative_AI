### Qdrant Vector Store (langchain_qdrant)

Qdrant is an open-source, high-performance vector search engine written in Rust.

It supports rich payload JSON filtering, in-memory execution, local disk storage, and cluster deployments.

### Deployment Modes

* **In-Memory**: location=":memory:" for rapid testing.
* **Disk Storage**: path="./qdrant_db" for local file persistence.
* **Remote / Cloud**: url="http://localhost:6333" for Docker containers or Qdrant Cloud.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_qdrant import QdrantVectorStore
from langchain_google_genai import GoogleGenerativeAIEmbeddings
# from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
from qdrant_client import QdrantClient
from qdrant_client.http import models

load_dotenv(find_dotenv())

# Initialize Google Generative AI embeddings (commented out HuggingFaceEmbeddings)
# embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")

docs = [
    Document(page_content="Qdrant is written in Rust for optimal speed and memory safety.", metadata={"lang": "rust", "year": 2021}),
    Document(page_content="Payload filtering allows querying metadata without payload penalties.", metadata={"lang": "python", "year": 2023}),
    Document(page_content="Vector indices use HNSW graphs for fast similarity retrieval.", metadata={"lang": "cpp", "year": 2020})
]

# Initialize Qdrant in-memory instance
qdrant = QdrantVectorStore.from_documents(
    documents=docs,
    embedding=embeddings,
    location=":memory:",
    collection_name="qdrant_demo"
)
print("Initialized Qdrant vector store in-memory.")

# Perform similarity search
results = qdrant.similarity_search("Rust memory safety", k=1)
print(f"Similarity Search Result: {results[0].page_content}")

# Advanced Qdrant Payload Filter
rust_filter = models.Filter(
    must=[
        models.FieldCondition(
            key="metadata.lang",
            match=models.MatchValue(value="rust")
        )
    ]
)
filtered_results = qdrant.similarity_search("vector search engine", filter=rust_filter, k=1)
print(f"Filtered Result (lang=rust): {filtered_results[0].page_content}")
```
