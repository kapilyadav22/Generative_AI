### Pinecone Cloud Vector Database (langchain_pinecone)

Pinecone is a fully managed, cloud-native vector database designed for enterprise production workloads.

It eliminates infrastructure management and provides real-time upserts, multi-tenant namespace partitioning, and low-latency similarity retrieval.

### Key Parameters & Concepts

* PINECONE_API_KEY: Authentication key loaded from environment.
* index_name: Target Pinecone index created in Pinecone Cloud Console.
* namespace: Optional logical partition for multi-tenant data isolation.
* Serverless Indexing: Auto-scaling cloud architecture with zero server management.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_pinecone import PineconeVectorStore
from langchain_google_genai import GoogleGenerativeAIEmbeddings
# from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document

load_dotenv(find_dotenv())

# Initialize Google Generative AI embeddings (commented out HuggingFaceEmbeddings)
# embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")

pinecone_api_key = os.getenv("PINECONE_API_KEY")
index_name = os.getenv("PINECONE_INDEX_NAME", "langchain-demo")

if pinecone_api_key:
    docs = [
        Document(page_content="Pinecone manages cloud vector indexing automatically.", metadata={"tenant": "corp_a"}),
        Document(page_content="Namespaces isolate data between users in a single index.", metadata={"tenant": "corp_b"})
    ]
    
    try:
        vectorstore = PineconeVectorStore.from_documents(
            documents=docs,
            embedding=embeddings,
            index_name=index_name,
            namespace="production-demo"
        )
        print("Successfully connected and indexed documents into Pinecone cloud.")
        results = vectorstore.similarity_search("cloud vector indexing", k=1)
        print(f"Pinecone Search Result: {results[0].page_content}")
    except Exception as e:
        print(f"Could not connect to Pinecone index '{index_name}': {e}")
else:
    print("PINECONE_API_KEY environment variable not found.")
    print("To execute Pinecone Cloud vector search, set PINECONE_API_KEY in your .env file.")
    print("\nAPI Code Structure:")
    print('vectorstore = PineconeVectorStore.from_documents(')
    print('    documents=docs,')
    print('    embedding=embeddings,')
    print('    index_name="my-pinecone-index",')
    print('    namespace="tenant-a"')
    print(')')
```
