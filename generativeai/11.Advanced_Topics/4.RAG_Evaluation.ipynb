# RAG Quality Metrics & Evaluation

Demonstrates evaluating RAG pipeline performance across three core quality pillars: Faithfulness, Answer Relevance, and Context Precision.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv(find_dotenv())
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")

question = "What is the distance between the Earth and the Moon?"
retrieved_context = "The average distance from the Earth to the Moon is approximately 384,400 kilometers (238,855 miles)."
generated_answer = "The Moon is located roughly 384,400 km away from Earth."

eval_prompt = f"""
Evaluate the generated answer based on context:
Context: {retrieved_context}
Question: {question}
Answer: {generated_answer}

Rate Faithfulness (1-5) and Answer Relevance (1-5). Provide brief reasoning.
"""

evaluation = llm.invoke(eval_prompt)
print(f"=== RAG Quality Evaluation Output ===\n{evaluation.content}")
```
