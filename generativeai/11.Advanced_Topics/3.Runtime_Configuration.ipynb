# Dynamic Runtime Configuration & Model Alternatives

Demonstrates configurable_fields and configurable_alternatives to dynamically switch models or parameters at runtime via config={'configurable': ...}.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv(find_dotenv())

primary_model = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
alternative_model = ChatGoogleGenerativeAI(model="gemini-3.6-flash")

# Configure alternative model at runtime
configurable_model = primary_model.configurable_alternatives(
    ConfigurableField(id="model_choice"),
    default_key="gemini_3_6",
    gemini_2_0=alternative_model
) if hasattr(primary_model, 'configurable_alternatives') else primary_model.with_fallbacks([alternative_model])

res = configurable_model.invoke("State the law of conservation of energy in one sentence.")
print(f"Configurable Model Response: {res.content}")
```
