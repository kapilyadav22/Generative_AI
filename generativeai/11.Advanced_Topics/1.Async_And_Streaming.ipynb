# Async Programming & Real-Time Token Streaming

Demonstrates asynchronous execution (ainvoke) and real-time token streaming (astream) using ChatGoogleGenerativeAI (gemini-3.6-flash).

```python
import asyncio
import os
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv(find_dotenv())
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")

# 1. Async Invoke (ainvoke)
async def run_async_demo():
    print("=== 1. Testing Async Invoke (ainvoke) ===")
    res = await llm.ainvoke("Name three major oceans on Earth.")
    print(f"Async Result: {res.content}\n")
    
    print("=== 2. Testing Token-by-Token Streaming (astream) ===")
    print("Streaming output: ", end="", flush=True)
    async for chunk in llm.astream("Write a 2-line poem about space."):
        print(chunk.content, end="", flush=True)
    print("\n")

asyncio.run(run_async_demo())
```
