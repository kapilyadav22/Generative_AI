# Response Caching (InMemoryCache & SQLiteCache)

Demonstrates using InMemoryCache and SQLiteCache to store LLM outputs, reducing latency and saving API costs on repeated queries.

```python
import time
import os
from dotenv import load_dotenv, find_dotenv
from langchain_community.cache import InMemoryCache
from langchain_core.globals import set_llm_cache
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv(find_dotenv())

# Enable In-Memory LLM Caching
set_llm_cache(InMemoryCache())
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")

query = "What is the speed of light in kilometers per second?"

# First call (Uncached - triggers LLM API)
t0 = time.time()
res1 = llm.invoke(query)
t1 = time.time()
print(f"First Call (Uncached API): Time = {t1 - t0:.3f}s | Response: {res1.content[:50]}...")

# Second call (Cached - returns instantly from memory)
t2 = time.time()
res2 = llm.invoke(query)
t3 = time.time()
print(f"Second Call (Cached Memory): Time = {t3 - t2:.3f}s | Response: {res2.content[:50]}...")
```
