# Retrievers in LangChain: Concepts & Architecture

A **Retriever** in LangChain is a lightweight, runnable interface designed to accept an un-embedded text query string as input and return a list of relevant Document objects.

Retrievers sit at the core of Retrieval-Augmented Generation (RAG) pipelines, acting as the bridge between raw user prompts and knowledge sources.

### VectorStore vs. Retriever: Key Differences

1. **VectorStore**: Manages physical vector data storage, persistent indexing (HNSW, IVF), embedding generation, and CRUD operations (create, read, update, delete).
2. **Retriever**: A generic LCEL (LangChain Expression Language) Runnable component that wraps a search algorithm or underlying store. It focuses purely on prompt-in, document-out search execution without exposing storage management APIs.

### Standard Search Types & Parameters Comparison

| Search Type | Syntax | Key Parameters | Strategy & Use Case |
|---|---|---|---|
| **similarity** | search_type="similarity" | k, filter | Standard top-k nearest neighbor semantic search. |
| **similarity_score_threshold** | search_type="similarity_score_threshold" | score_threshold, k, filter | Filters out document chunks below minimum confidence threshold. |
| **mmr** | search_type="mmr" | k, fetch_k, lambda_mult | Maximal Marginal Relevance; balances similarity with result diversity. |
